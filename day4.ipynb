#### loc and iloc in panda

```python
import pandas as pd
data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Emma"],
    "Age": [23, 25, 22, 24, 26],
    "Department": ["HR", "IT", "Finance", "IT", "Marketing"],
    "Salary": [40000, 50000, 45000, 52000, 48000]
}
df=pd.DataFrame(data)
print(df)
```

```python
df.loc[[1,3]]
```

```python
df.loc[2,"Name"]
```

```python
df.loc[2,["Name","Salary"]]
```

```python
df.loc[[1,3],["Name","Department"]]
```

```python
df.loc[1:3]
```

```python
df.loc[df["Salary"]>45000]
```

```python
df.loc[df["Salary"] > 45000, ["Name", "Salary"]]
```

```python
df.loc[2,"Salary"]=47000
df
```

```python
import pandas as pd

data = {
    "Name": ["A", "B", "C", "D", "E", "F"],
    "Age": [23, 45, 31, 22, 35, 28],
    "Salary": [40000, 80000, 50000, 30000, 90000, 60000],
    "Department": ["HR", "IT", "Finance", "HR", "IT", "Finance"]
}

df = pd.DataFrame(data)

# custom index
df.index = ["emp1", "emp2", "emp3", "emp4", "emp5", "emp6"]

print(df)
```

```python
df.iloc[0:3]
```

```python
df.iloc[2:5,1:3]
```

```python
df.iloc[:, 0:2]
```

```python
df.loc["emp1":"emp3"]
```
