```python
import pandas as pd
data = {
    "Department": ["HR", "IT", "IT", "HR", "Finance", "IT", "Finance"],
    "Employee": ["A", "B", "C", "D", "E", "F", "G"],
    "Salary": [40000, 60000, 65000, 45000, 70000, 62000, 72000],
    "Experience": [2, 5, 6, 2, 8, 4, 8],
    "City":["Bangalore","Chennai","Kochi","Kochi","Chennai","Kochi","Mumbai"]
}

df1 = pd.DataFrame(data)
df1
```

```python
df1["Dept_Avg_Salary"] = df1.groupby("Department")["Salary"].transform("mean")
df1
```

```python
df1["Salary_Diff"] = df1["Salary"] - df1["Dept_Avg_Salary"]
df1
```

```python
df1["Salary_Percent"]=df1["Salary"]/df1.groupby("Department")["Salary"].transform("sum")
```

```python
df1
```

```python
df1["Dpt_avg_Salary"]=df1.groupby("Department")["Salary"].transform("mean")
```

```python
df1
```

```python
df1["count_emp_Dept"] =df1.groupby("Department")["Employee"].transform("count")
```

```python
df1["max_Salary_per_Dpt"]= df1.groupby("Department")["Salary"].transform("max")
df1
```
